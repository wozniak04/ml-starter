## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python

import pandas as pd


df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')


df_filtered = df[df['impressions_90d'] > 0].copy()


total_pages = len(df_filtered)
pages_with_ai_sessions = df_filtered[df_filtered['ai_sessions_90d'] > 0]
num_pages_with_ai_sessions = len(pages_with_ai_sessions)


percentage_with_ai = (num_pages_with_ai_sessions / total_pages) * 100
total_ga_sessions = df_filtered['sessions_90d'].sum()
total_ai_sessions = df_filtered['ai_sessions_90d'].sum()

print(f"Total pages in the dataset: {total_pages}")
print(f"Pages with at least one AI session: {num_pages_with_ai_sessions} ({percentage_with_ai:.2f}%)")
print(f"Total sessions from search engines (90d): {total_ga_sessions:,.0f}")
print(f"Total sessions from AI (90d): {total_ai_sessions:,.0f}")
if total_ga_sessions > 0:
    ai_to_ga_ratio = (total_ai_sessions / total_ga_sessions) * 100
    print(f"Percentage of AI sessions relative to search engine sessions: {ai_to_ga_ratio:.2f}%")

print("\nAlthough AI traffic is rare, this rarity is what makes it interesting. The pages that acquire it are outliers. Analyzing these exceptions can reveal patterns that will help other pages gain this new kind of visibility.")
```
