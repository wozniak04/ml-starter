## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
import pandas as pd

# --- My success metric is Precision@50 ---
#
# I need a way to check if my ranked list is any good.
# This function will look at the top 50 pages my system recommends
# and tell me what percentage of them actually had any AI traffic.
# It's a simple way to see if I'm saving the editor's time.

def calculate_precision_at_k(df, k=50, score_col='score', label_col='is_positive'):
    """Calculates precision at K for a ranked dataframe."""
    top_k = df.sort_values(by=score_col, ascending=False).head(k)
    precision = top_k[label_col].mean()
    return precision

# Load data
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')
df_filtered = df[df['impressions_90d'] > 0].copy()

# A "positive" page is one that had any AI traffic
df_filtered['is_positive'] = df_filtered['ai_sessions_90d'] > 0

# --- Let's create a really simple baseline to compare against ---
# My simple idea: maybe pages with more search impressions get more AI traffic?
# I'll use this as my 'score' for now.
df_filtered['score'] = df_filtered['impressions_90d']

# Calculate the metric for this simple idea
p_at_50 = calculate_precision_at_k(df_filtered, k=50, score_col='score', label_col='is_positive')

print(f"My simple baseline (ranking by impressions) gets a Precision@50 of: {p_at_50:.2%}")
print("\nThis is the score my real ML model has to beat.")
```
